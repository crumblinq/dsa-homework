````python
import csv
with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))
print(len(rows), "строк")
````

## [База]
1. Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной).
2. Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.
3. Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.
4. Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`.

````python
# База
# 1 Напишите функции to_num(s) и clean_city(s) (6 городов, латиница и разный регистр → кириллица с заглавной)
def to_num(s):
    return float(s.replace(' ', ''))
   
def clean_city(s):
    return s.strip().lower().title().replace('Almaty', 'Алматы').replace('Astana', 'Астана').replace('Shymkent', 'Шымкент')
````

````python
for r in rows:
    r['city'] = clean_city(r['city'])
    r['amount'] = to_num(r['amount'])
````

````python
# проверка
print([r['city'] for r in rows[:50]])
````

````python
# 2 Сколько заказов в каждой категории (category)? Отсортируйте по убыванию.
count =  {}
for r in rows:
    if r['category'] in count:
        count[r['category']] += 1
    else:
        count[r['category']] = 1
print(count)
````

````python
from collections import Counter
print(Counter(count).most_common())
````

````python
# 3 Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.
revenue = {}
paid_count = {}

for r in rows:
    if r['status'] == 'оплачен':
        revenue[r['category']] = revenue.get(r['category'], 0) + r['amount']
        paid_count[r['category']] = paid_count.get(r['category'], 0) + 1
````

````python
print(f'{"категория":<11} | {"выручка":>10} | {"средний чек"}')
for c in revenue:
    print(f'{c:<11} | {revenue[c]:>10} | {revenue[c] / paid_count[c]:.2f}')
````

````python
# 4 Сколько заказов в каждом месяце? Месяц — это order_date[:7]
order_per_month_count = {}

for r in rows:
    month = r['order_date'][:7]
    if month in order_per_month_count:
        order_per_month_count[month] += 1
    else:
        order_per_month_count[month] = 1
print(order_per_month_count)
````

## [Стандарт]
5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.
6. Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.
7. Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.
8. Сколько **уникальных** клиентов в каждом городе? Подсказка: словарь множеств.

````python
# Стандарт
# 5 Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.
revenue_per_month = {}

for r in rows:
    if r['status'] == 'оплачен':
        month = r['order_date'][:7]
        if month in revenue_per_month:
            revenue_per_month[month] += r['amount']
        else:
            revenue_per_month[month] = r['amount']
print(revenue_per_month)
````

````python
sep = revenue_per_month['2026-09']
oct = revenue_per_month['2026-10']

print(f'Изменение: {((oct - sep) / sep) * 100:.1f}%')
````

````python
# 7 Топ-5 клиентов (customer_id) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.
customer_revenue = {}

for r in rows:
    if r['status'] == 'оплачен':
        customer_revenue[r['customer_id']] = customer_revenue.get(r['customer_id'], 0) + r['amount']
````

````python
print(Counter(customer_revenue).most_common(5))
````
