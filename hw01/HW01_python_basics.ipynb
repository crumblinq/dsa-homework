````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# База
# 1 Приведите все города к единому виду ("Almaty", "Astana", …) и сохраните в новый список cities_clean
cities_clean = [c.strip().title() for c in cities]
print(cities_clean)
````

````python
# 2 Преобразуйте amounts из строк в числа: список amounts_int
amounts_int = []
for a in amounts:
    amounts_int.append(int(a))
print(amounts_int)
````

````python
# 3 Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого)
print('Общее число заказов:', len(amounts_int))
print('Общая сумма заказов:', sum(amounts_int))
print('Средний чек:', round(sum(amounts_int) / len(amounts_int)))
````

````python
# 4 Сколько заказов не оплачено и на какую сумму?
not_paid = 0
total = 0

for p, a in zip(paid, amounts_int):
    if not p:
        not_paid += 1
        total += a
print(not_paid, total)
````

````python
# 5 Выведите таблицу в формате город | сумма | статус, используя zip и f-строку
print(f'{"город":<10} | {"сумма":>6} | {"статус"}')
for c, a, p in zip(cities_clean, amounts_int, paid):
    print(f'{c:<10} | {a:>6} | {p}')
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# Стандарт
# 6 Выручка (только оплаченные) по Алматы и по Астане отдельно
astana_total = 0
almaty_total = 0

for c, a, p in zip (cities_clean, amounts_int, paid):
    if c == 'Almaty' and p:
        almaty_total += a
    if c == 'Astana' and p:
        astana_total += a
print('Выручка в Алматы:', almaty_total)
print('Выручка в Аcтане:', astana_total)
````

````python
# 7 Номер (индекс) и город самого крупного оплаченного заказа
highest = 0
for a, p, c in zip(amounts_int, paid, cities_clean):
    if p and a > highest:
        highest = a
        highest_city = c
print('Самый крупный оплаченный заказ:', highest, highest_city)

for i, c in enumerate(cities_clean):
    if amounts_int[i] == highest:
        print(i, c)
````

````python
# 8 Доля заказов выходного дня (сб, вс) в процентах
weekends = 0
for d in days:
    if d == 'сб' or d == 'вс':
        weekends += 1
print(f'{weekends / len(days) * 100:.0f}%')
````

````python
# 9 Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. 
# С помощью while посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый

balance = 400000
saving = 15000
months = 0

while balance > 0:
    balance -= saving
    months += 1
    print(f'Месяц:', months, f'Отложил в этом месяце:', saving, f'Осталось накопить:', balance)
print(f'Студенту нужно копить {months} мес.')
````

````python
balance = 400000
saving = 15000
months = 0

while balance > 0:
    payment = saving + 1000 * months
    balance -= payment
    months += 1
    print(f'Месяц:', months, f'Отложил в этом месяце:', payment, f'Осталось накопить:', balance)
print(f'Студенту нужно копить {months} мес.')
````
